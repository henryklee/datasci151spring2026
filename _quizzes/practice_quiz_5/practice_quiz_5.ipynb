**Q1** Recode a numeric column

<font size = "3">

- Read in the file "movies.csv" from the `quiz_data` folder, saving it as a DataFrame.

- Recode the "year" variable to create a new variable called "time_period" with the following categories.

$\qquad$ ``` ["1941-1967","1968-2003","2004-onwards"] ```

- You must use the `pd.cut` function to recode this variable.

- Print the DataFrame to the screen to check your answers


````python
# your code here
# Read in the CSV file
movies = pd.read_csv("quiz_data/movies.csv")

# pd.cut() divides a numeric column into discrete bins.
# 'bins' defines the boundary edges of each interval.
#   - 1940 is the left edge (exclusive) so the first bin captures from 1941 onward
#   - 1967, 2003, and inf are the remaining edges
# 'labels' assigns a human-readable string to each resulting bin.
#   - There must be exactly one fewer label than there are bin edges.
movies["time_period"] = pd.cut(
    movies["year"],
    bins=[1940, 1967, 2003, float("inf")],
    labels=["1941-1967", "1968-2003", "2004-onwards"]
)

print(movies)
````

**Q2** Aggregate and Sort

<font size = "3">

- Continue to use the DataFrame from the previous question.

- Do the following in a **single line of code** using chaining:

    - Group the data by "country"
    
    - Aggregate the following statistics: average runtime, maximum runtime, average IMDB score, minimum IMDB score

    - Sort by average IMDB score in **descending order**

- Print the resulting DataFrame to the screen

**Note:** A "single line of code" can continue on to the next line if you use a backslash or wrap the line within a pair of parentheses.

````python
# your code here
# This is one chained expression broken across lines using parentheses.
#
# .groupby("country")  — splits the DataFrame into one group per country
#
# .agg({...})  — applies multiple aggregation functions at once.
#   Each key is a column name, and each value is a list of functions to apply.
#   "mean" → average, "max" → maximum, "min" → minimum
#
# .sort_values(("imdb_score", "mean"), ascending=False)
#   — sorts by the (imdb_score, mean) column in the resulting MultiIndex columns,
#     with the highest average IMDB score first.

result_q2 = (
    movies
    .groupby("country")
    .agg(
        avg_runtime=("runtime(minutes)", "mean"),
        max_runtime=("runtime(minutes)", "max"),
        avg_imdb=("imdb_score", "mean"),
        min_imdb=("imdb_score", "min")
    )
    .sort_values("avg_imdb", ascending=False)
)

print(result_q2)
````

**Q4** Aggregate and query

<font size = "3">

- Read in the file "sprint_results.csv" from the `quiz_data` folder, saving it as a DataFrame.

- Do the following in a single line using chaining:

    - Extract the subset with constructorId equal to 51.

    - Aggregate the maximum number of laps and average number of laps for each driver in the subset.

    - Extract the subset of these drivers with maximum # of laps $\geq 20$. 


````python
# your code here
sprint = pd.read_csv("quiz_data/sprint_results.csv")

# .query("constructorId == 51")
#   — keep only rows where constructorId is 51.
#
# .groupby("driverId")
#   — group those filtered rows by each unique driver.
#
# .agg(...)
#   — for each driver, calculate:
#       max_laps: the most laps they completed in any single sprint
#       avg_laps: their average laps across all sprints
#
# .query("max_laps >= 20")
#   — after aggregation, keep only drivers whose maximum laps was 20 or more.
#   We can query on the new aggregated column names directly.

result_q4 = (
    sprint
    .query("constructorId == 51")
    .groupby("driverId")
    .agg(
        max_laps=("laps", "max"),
        avg_laps=("laps", "mean")
    )
    .query("max_laps >= 20")
)

print(result_q4)
````

**Q5** Calculate race duration

<font size = "3">

- Read in the file "results_cleaned.csv" from the `quiz_data` folder, saving it as a DataFrame.

- The "milliseconds" column shows the race duration for each result measured in milliseconds. Create a new column called "race_duration_minutes" where we convert the race duration to minutes by dividing each millisecond value by 60,000 (which equals $6 \times 10^4$). Equivalently, you can multiply each millisecond value by $6 \times 10^{-4}$

- Group the data by **both** "raceId" and "constructorId" and calculate the average race duration in minutes for each group.

- Sort the data in descending order and then extract the raceId/constructorId pairs with the 5 slowest average times.


````python
# your code here
results = pd.read_csv("quiz_data/results_cleaned.csv")

# Divide the milliseconds column by 60,000 to convert to minutes.
# This creates a brand new column on the DataFrame.
results["race_duration_minutes"] = results["milliseconds"] / 60_000

# .groupby(["raceId", "constructorId"])
#   — groups by the combination of both columns, so each group is one
#     constructor's performance in one specific race.
#
# ["race_duration_minutes"].mean()
#   — computes the average duration in minutes for each raceId/constructorId pair.
#
# .sort_values(ascending=False)
#   — puts the slowest (longest) average times at the top.
#
# .head(5)
#   — returns only the 5 slowest pairs.

result_q5 = (
    results
    .groupby(["raceId", "constructorId"])["race_duration_minutes"]
    .mean()
    .sort_values(ascending=False)
    .head(5)
)

print(result_q5)
````

**Q6** Aggregate custom statistic

<font size = "3">

- Continue to use the DataFrame corresponding to the data in "results_cleaned.csv".

- When a driver finishes 1st, 2nd, or 3rd it is known as a "podium finish". Write a function that takes in a Pandas Series and returns the number of entries $\leq 3$.

- Group by "driverId" and then aggregate the following statistics: the average of the "points" column, and the number of podium finishes in the "positionOrder" column - use your custom function for this second statistic.

- Merge the aggregated data back into the original DataFrame.

````python
# your code here
# Define a custom aggregation function.
# It receives an entire Pandas Series (one group's worth of positionOrder values)
# and returns a single number: how many entries are 3 or below (podium finishes).
# (series <= 3) produces a boolean Series; .sum() counts the True values.
def count_podiums(series):
    return (series <= 3).sum()

# .groupby("driverId")
#   — one group per driver across all their race results.
#
# .agg({...})
#   — apply different functions to different columns:
#       "points" → mean (average points per race)
#       "positionOrder" → our custom count_podiums function
#
# The result is a new DataFrame indexed by driverId with two columns.
agg_df = (
    results
    .groupby("driverId")
    .agg(
        avg_points=("points", "mean"),
        podium_finishes=("positionOrder", count_podiums)
    )
    .reset_index()  # turns driverId back into a regular column so we can merge on it
)

# pd.merge() joins two DataFrames on a shared key column.
# on="driverId" — the column to match rows between the two DataFrames.
# how="left"    — keep ALL rows from `results` (the left DataFrame);
#                 attach the aggregated stats wherever driverId matches.
results_merged = pd.merge(results, agg_df, on="driverId", how="left")

print(results_merged[["driverId", "points", "positionOrder", "avg_points", "podium_finishes"]].head(10))
````
